# Merge

```python
input_path = 'TiN_R10_q0M1.csv'
output_path = 'TiN_R5_q0M1_Ko.csv'

with open(input_path, 'r') as f:
    lines = f.readlines()

# 各行をfloatのリストに変換
rows = []
for line in lines:
    vals = [float(v) for v in line.strip().split(',')]
    rows.append(vals)

new_rows = []

# 1行目はそのまま
new_rows.append(rows[0])

# 2+3, 4+5, 6+7, 8+9 行目をペアで足し合わせる
for i in range(1, len(rows)-1, 2):
    combined = [a + b for a, b in zip(rows[i], rows[i+1])]
    new_rows.append(combined)

# 出力
with open(output_path, 'w') as f:
    for row in new_rows:
        f.write(','.join(f'{v:.6e}' for v in row) + '\n')
```

# Cutting

```python
input_path = 'TiN_R5_q0M1_Ko.csv'
output_path = 'TiN_R5_q0M1_Ko_thresholded.csv'

with open(input_path, 'r') as f:
    lines = f.readlines()

# 各行をfloatのリストに変換
rows = []
for line in lines:
    vals = [float(v) for v in line.strip().split(',')]
    rows.append(vals)

# 各行の最大値の1e-4倍より小さい値を0にする
new_rows = []
for row in rows:
    row_max = max(row)
    threshold = row_max * 1e-2
    new_row = [v if v >= threshold else 0.0 for v in row]
    new_rows.append(new_row)

# 出力
with open(output_path, 'w') as f:
    for row in new_rows:
        f.write(','.join(f'{v:.6e}' for v in row) + '\n')
```

```python
np.sum(new_rows)/np.sum(rows)
```

# Plotting

```python
norm = 1e20 / 3.17e-8

import numpy as np
import matplotlib.pyplot as plt

input_path = 'TiN_R5_q0M1_Ko.csv'

# データ読み込み
rows = []
legends = ['[0.2, 0.3] eV', '[0.3, 0.5] eV', '[0.5, 0.7] eV', '[0.7, 0.9] eV', '[0.9, 1.1] eV']
with open(input_path, 'r') as f:
    for line in f:
        vals = [float(v) * TiNexp/sec / 4.95e-9/norm for v in line.strip().split(',')]
        rows.append(vals)

x = np.arange(len(rows[0]))

# 折れ線グラフ
fig, ax = plt.subplots(figsize=(8, 6))
plt.rcParams['ytick.labelsize'] = 20
plt.rcParams['xtick.labelsize'] = 20
for i, row in enumerate(rows):
    #ax.scatter(x, row, label=f'{legends[i]}', s = 1)
    ax.plot(x, row, label=f'{legends[i]}', lw = 3)
ax.set_xlabel(r'$v_{\text{min}}$ [km s$^{-1}$]', fontsize = 25)
ax.set_ylabel(r'$\mathcal{R}_{\left[E_i, E_{i + 1}\right]}$ [$10^{20}$ kg$^{-1}$]', fontsize = 25)
ax.set_title(r'MKID: $m=10$ MeV', fontsize = 25)
#ax.set_yscale('log')
#ax.set_xscale('log')
ax.legend(fontsize = 25, frameon = False)

#plt.gca().xaxis.set_minor_locator(LogLocator(base=10.0, subs=np.arange(1.0, 10) * 0.1, numticks=20))

plt.tick_params(axis='both', which='major', labelsize=25)
plt.tick_params(axis='both', which='minor', labelsize=25)

plt.minorticks_on()
plt.xticks( np.arange(0, 350, 50))

ax.set_xlim(0, 300)
ax.set_ylim(0, 9)
plt.tight_layout()
#plt.savefig('lineplot.png', dpi=150)
#plt.close()

# 散布図
#fig, ax = plt.subplots()
#for i, row in enumerate(rows):
#    ax.scatter(x, row, label=f'Row {i+1}', s=1)
#ax.set_xlabel('Column Index')
#ax.set_ylabel('Value')
##ax.set_title('Scatter Plot')
#ax.legend()
##ax.set_xscale('log')
#ax.set_xlim(0, 200)
#plt.tight_layout()
#plt.grid(True, which="both", ls="--")
plt.savefig('curlyR_MKID_10MeV_q=0_5bins_thresholded.pdf', dpi=150)
##plt.close()
```

# Refined exposure

```python
input_path = 'TiN_R5_q0M1_Ko_thresholded.csv'
output_path = 'TiN_R5_q0M1_Ko_thresholded_refined_exposure.csv'

with open(input_path, 'r') as f:
    lines = f.readlines()

# 各行をfloatのリストに変換
rows = []
for line in lines:
    vals = [float(v) * TiNexp for v in line.strip().split(',')]
    rows.append(vals)

# 各行の最大値の1e-4倍より小さい値を0にする
new_rows = []
for row in rows:
    row_max = max(row)
    threshold = row_max * 1e-2
    new_row = [v if v >= threshold else 0.0 for v in row]
    new_rows.append(new_row[60:259])

# 出力
with open(output_path, 'w') as f:
    for row in new_rows:
        f.write(','.join(f'{v:.6e}' for v in row) + '\n')
```
