# Merge

```python
input_path = 'Al_R10_q0M1.csv'
output_path = 'Al_R5_q0M1_Ko.csv'

with open(input_path, 'r') as f:
    lines = f.readlines()

# 各行をfloatのリストに変換
rows = []
for line in lines:
    vals = [float(v) for v in line.strip().split(',')]
    rows.append(vals)

new_rows = []

# 1行目はそのまま
new_rows.append(rows[0])

# 2+3, 4+5, 6+7, 8+9 行目をペアで足し合わせる
for i in range(1, len(rows)-1, 2):
    combined = [a + b for a, b in zip(rows[i], rows[i+1])]
    new_rows.append(combined)

# 出力
with open(output_path, 'w') as f:
    for row in new_rows:
        f.write(','.join(f'{v:.6e}' for v in row) + '\n')
```

```python
import numpy as np
from scipy.interpolate import UnivariateSpline
from scipy.integrate import quad

# ===== 定数 =====
sec = 1.51976e15
Alexp = 1.84056e49
norm = 1

legends = ['[0.1, 0.3] eV', '[0.3, 0.5] eV', '[0.5, 0.7] eV', '[0.7, 0.9] eV', '[0.9, 1.1] eV']

# ===== スプライン関数を作る関数 =====
# ===== スプライン関数を作る汎用関数 =====
def make_splines(input_path, xlim=200, plot_title='', save_path=None):
    """
    CSVを読み込み、各行（エネルギービン）に対して
    UnivariateSpline でスムーズな補間関数を返す。
 
    Returns
    -------
    splines : list of callable
        splines[i](v) で i番目のビンの値を返す（vはkm/s単位のインデックス）
    """
    rows = []
    with open(input_path, 'r') as f:
        for line in f:
            vals = [float(v) * Alexp
                    for v in line.strip().split(',')]
            rows.append(vals)
 
    splines = []
    for i, row in enumerate(rows):
        x = np.arange(len(row))
        y = np.array(row)
 
        # ゼロ除外してスプライン作成（ゼロ点は補間精度を落とすため）
        mask = y > 0
        if mask.sum() < 4:
            # データ点が少なすぎる場合はゼロ関数
            splines.append(lambda v, _y=y: np.zeros_like(np.atleast_1d(v), dtype=float))
            continue
 
        x_nz = x[mask]
        y_nz = y[mask]
 
        # s=0 で厳密補間、s>0 で平滑化（必要に応じて調整）
        spl = UnivariateSpline(x_nz, y_nz, s=0, ext=1)  # ext=1: 範囲外は0
        splines.append(spl)
 
    return splines


# ===== 積分して行列を作る関数 =====
def make_integral_matrix(splines, index_range):
    """
    Parameters
    ----------
    splines : list of 5 callables
        各エネルギービンのスプライン関数
    index_range : tuple (a, b)
        積分する区間のインデックス範囲（例: (4, 128)）
        [a, a+1], [a+1, a+2], ..., [b-1, b] の各区間で積分

    Returns
    -------
    matrix : np.ndarray, shape (5, b-a)
        matrix[i, j] = j番目の区間でのi番目のビンの積分値
    """
    a, b = index_range
    intervals = [(k, k+1) for k in range(a, b)]  # [a,a+1], [a+1,a+2], ..., [b-1,b]
    n_bins = len(splines)       # 5
    n_intervals = len(intervals)  # b - a

    matrix = np.zeros((n_bins, n_intervals))

    for i, spl in enumerate(splines):
        for j, (lo, hi) in enumerate(intervals):
            val, _ = quad(spl, lo, hi)
            matrix[i, j] = max(val, 0.0) * (km/cm)  # 負値をゼロにクリップ

        print(f"  bin {i+1}/5 done")

    return matrix / cm
```

```python
# ===== 実行例 =====
if __name__ == '__main__':
    import os

    # --- m = 1 GeV の例 ---
    input_path = 'Al_R5_q0M3_Ko.csv'

    if not os.path.exists(input_path):
        print(f"ファイルが見つかりません: {input_path}")
        print("動作確認用にダミーデータで実行します。")

        # ダミーデータで動作確認
        def dummy_spline(v):
            v = np.atleast_1d(v)
            return np.exp(-v / 30.0) * (v > 4)

        splines = [dummy_spline] * 5
    else:
        print(f"Splining: {input_path}")
        splines = make_splines(input_path)

    # (4, 128) を渡す → [4,5], [5,6], ..., [127,128] の124区間で積分
    index_range = (4, 128)
    a, b = index_range
    print(f"\nPerforming integration: {index_range} → {b - a} intervals")

    matrix = make_integral_matrix(splines, index_range)

    print(f"\nMatrix size: {matrix.shape}")  # (5, 124)

    # CSV保存
    output_path = 'integral_matrix_1GeV.csv'
    np.savetxt(output_path, matrix, delimiter=',', fmt='%.6e')
    print(f"\nSuccesfully saved: {output_path}  shape={matrix.shape}")
```
