# Cutting

```python
input_path = 'Al_R10_q0M1.csv'
output_path = 'Al_R10_q0M1_Ko_thresholded.csv'

with open(input_path, 'r') as f:
    lines = f.readlines()

# 各行をfloatのリストに変換
rows = []
for line in lines:
    vals = [float(v) for v in line.strip().split(',')]
    rows.append(vals)

# 各行の最大値の1e-4倍より小さい値を0にする
new_rows = []
for row in rows:
    row_max = max(row)
    threshold = row_max * 1e-2
    new_row = [v if v >= threshold else 0.0 for v in row]
    new_rows.append(new_row)

# 出力
with open(output_path, 'w') as f:
    for row in new_rows:
        f.write(','.join(f'{v:.6e}' for v in row) + '\n')
```

# Refined exposure

```python
input_path = 'Al_R10_q0M1_Ko_thresholded.csv'
output_path = 'Al_R10_q0M1_Ko_thresholded_refined_exposure.csv'

with open(input_path, 'r') as f:
    lines = f.readlines()

# 各行をfloatのリストに変換
rows = []
for line in lines:
    vals = [float(v) * Alexp * 3.336e-6 /cm for v in line.strip().split(',')]
    rows.append(vals)

# 各行の最大値の1e-4倍より小さい値を0にする
new_rows = []
for row in rows:
    row_max = max(row)
    threshold = row_max * 1e-2
    new_row = [v if v >= threshold else 0.0 for v in row]
    new_rows.append(new_row[44:535])

# 出力
with open(output_path, 'w') as f:
    for row in new_rows:
        f.write(','.join(f'{v:.6e}' for v in row) + '\n')
```
